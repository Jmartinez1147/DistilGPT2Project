```python
import torch
import transformers

print("Torch:", torch.__version__)
print("Transformers:", transformers.__version__)
```

```python
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM
)

model_id = "distilbert/distilgpt2"

tokenizer = AutoTokenizer.from_pretrained(
    model_id
)

model = AutoModelForCausalLM.from_pretrained(
    model_id
)

print(
    "Loaded:",
    model_id
)
```

```python
text = "machine learning is powerful"

tokens = tokenizer.tokenize(text)

ids = tokenizer.encode(text)

decoded = tokenizer.decode(ids)

print("Tokens:", tokens)

print("Token IDs:", ids)

print("Decoded:", decoded)
```

```python
word = "unbelievable"

print(
    tokenizer.tokenize(word)
)
```

## Token Inspection

The tokenizer converts text into tokens and token IDs.

Some words are represented by multiple tokens instead of a single token.

The decoded output reconstructs text from token IDs.

```python
torch.manual_seed(42)

prompt = "Students learn best when"

inputs = tokenizer(
    prompt,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=30,
    do_sample=True,
    temperature=0.7
)

print(
    tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )
)
```

```python
torch.manual_seed(42)

outputs = model.generate(
    **inputs,
    max_new_tokens=30,
    do_sample=True,
    temperature=1.2
)

print(
    tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )
)
```

## Sampling Experiment

The same prompt was generated multiple times using different random seeds.

Changing the random seed produced different outputs even when the prompt remained the same.

Increasing temperature increased randomness and produced more varied text.

Higher temperatures may increase creativity but can also reduce consistency and reliability.

```python
prompt1 = "Students learn best when"

prompt2 = "Students learn best when they actively practice concepts and receive feedback."

for prompt in [prompt1, prompt2]:

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=30,
        do_sample=True,
        temperature=0.7
    )

    print("\nPROMPT:")
    print(prompt)

    print("\nOUTPUT:")

    print(
        tokenizer.decode(
            outputs[0],
            skip_special_tokens=True
        )
    )
```

## Context Experiment Results

The wording and amount of context affected the generated text.

The shorter prompt produced a more general continuation.

The longer prompt guided the model toward a more specific continuation.

This demonstrates that transformers use surrounding context when generating text.

## Failure or Limitation

One surprising behavior was that some generated text did not always stay focused on the original topic.

The model occasionally produced repetitive or unusual wording.

In a customer-facing product, this could create confusing or inaccurate responses.

Generated text should be reviewed carefully because the model can produce incorrect information even when the text sounds confident.
